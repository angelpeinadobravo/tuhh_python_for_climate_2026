## 2. Variables: giving names to values

A **variable** stores a value under a name, with `=`. Use the name later instead of the value.
`print(...)` shows a value, and you can print several things in one cell.

```python
temperature_celsius = 15.3
temperature_kelvin = temperature_celsius + 273.15

print(temperature_celsius)
print(temperature_kelvin)
```

Text goes between quotes and is called a **string**. An *f-string* (`f"..."`) puts variables into a text with `{}`:

```python
city = "Hamburg"
print(f"In {city} it is {temperature_celsius} °C, that is {temperature_kelvin:.1f} K.")
```

**Try it:** change the city and the temperature above, and run the two cells again.

## 3. Lists: many values in one variable

A **list** keeps several values in order, between `[ ]`. Python counts **from 0**: the first element is `[0]`.

```python
# example values, made up for this exercise
daily_temperatures = [12.1, 13.4, 15.0, 14.2, 11.8]

print(daily_temperatures[0])  # first element
print(daily_temperatures[-1])  # last element
print(len(daily_temperatures))  # number of elements
```

## 4. Functions: reusable pieces of code

A **function** takes inputs, does something, and gives back a result with `return`.
Python already has many functions (`print`, `len`, `max`, ...), and you can write your own with `def`.
The lines inside the function are **indented** (4 spaces): that is how Python knows they belong to it.

```python
def celsius_to_kelvin(celsius):
    return celsius + 273.15


print(celsius_to_kelvin(0))
print(max(daily_temperatures))
```

## 5. Packages: code written by others

Python's strength is its **packages**. You load them with `import`. The workshop container already has all the
packages we need. `numpy` (short `np`) works with numbers and arrays:

```python
import numpy as np

temperatures = np.array(daily_temperatures)
print(temperatures.mean())  # average
print(celsius_to_kelvin(temperatures))  # works on all values at once
```

## 6. Your first plot

`matplotlib` makes plots. Every plot has a **figure** (`fig`) with one or more **axes** (`ax`) to draw on.

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]

fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(days, daily_temperatures, marker="o")
ax.set_xlabel("day")
ax.set_ylabel("temperature / °C")
ax.set_title("Example temperatures")
plt.show()
```

## 7. Your region on the globe

Now the picture for **your page on the workshop webpage**. `cartopy` draws maps.

1. Change `my_name`, `region_name`, `lon` and `lat` to a region of the world you are interested in.
   Longitude goes from -180 (west) to 180 (east), latitude from -90 (south) to 90 (north).
   You can find them on any online map.
2. Run the cell. The map is shown below and saved as `my_region.png` next to your page.

The first map can take a minute: cartopy downloads the coastlines once.

```python
import cartopy.crs as ccrs
import cartopy.feature as cfeature

my_name = "Firstname Lastname"
region_name = "Hamburg"
lon, lat = 10.0, 53.55  # longitude, latitude in degrees

# the picture of your page on the webpage (pages/01_setup/my_page.rst)
file_path = "/app/pages/01_setup/my_region.png"

fig = plt.figure(figsize=(5, 5))
# a globe seen from above your region
ax = plt.axes(projection=ccrs.Orthographic(central_longitude=lon, central_latitude=lat))
ax.set_global()
ax.add_feature(cfeature.OCEAN, color="lightblue")
ax.add_feature(cfeature.LAND, color="wheat")
ax.coastlines(linewidth=0.5)
ax.gridlines(linewidth=0.3, color="gray")
# transform=ccrs.PlateCarree(): lon and lat are plain degrees
ax.plot(lon, lat, "o", color="red", markersize=10, transform=ccrs.PlateCarree())
ax.set_title(f"{region_name} · {my_name}")

fig.savefig(file_path, dpi=100, bbox_inches="tight")
plt.show()
print("Saved as", file_path)
```
